### Задание 1
Датасет для домашнего задания вы найдете в разделе “Материалы к домашнему заданию” к этому занятию.
Определите, какому фильму было выставлено больше всего оценок 5.0.

```python
import pandas as pd
```

```python
df_1 = pd.read_csv("movies.csv")
df_1.head()
```

```python
df_2 = pd.read_csv("ratings.csv")
df_2.head()
```

```python
# объединяем файлы
merged = pd.merge(df_2, df_1, on='movieId', how='left')
merged.head()
```

```python
# фильтруем по условию, чтобы рейтинг был 5.0
merged_filter_raring = merged[merged['rating'] == 5.0]
merged_filter_raring.head()
```

```python
# выделяем топ - фильм, у которого больше всего оценок от пользователей 5.0
# берем названия фильмов title
# и считаем количество уникальных значений .value_counts()
# .value_counts() сортирует результат по убыванию частоты: самое распространённое значение будет в первой строке
# выводим первую строку в списке .head(1)
# она и будет фильмом с максимальным количеством оценок
top = merged_filter_raring['title'].value_counts().head(1)
top
```

### Вывод по результатам решения:

```python
print(f'Больше всего оценок 5.0 выставлено фильму {top.index[0]}, количество оценок {top.iloc[0]}')
```
